### **Langkah 3: Latih Model**

```python
from sklearn.ensemble import RandomForestClassifier

# Membuat model Random Forest
rf = RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42
)

# Melatih model menggunakan data latih
rf.fit(X_tr, y_tr)

print("Model Random Forest berhasil dilatih.")
```

Pada tahap ini dibuat model Random Forest dengan jumlah 300. Parameter class_weight="balanced" digunakan untuk mengatasi ketidakseimbangan jumlah kelas pada dataset dengan memberikan bobot yang lebih besar pada kelas minoritas, yaitu pelanggan yang melakukan churn. Setelah model dilatih menggunakan data latih, model siap digunakan untuk melakukan prediksi pada data uji dan dievaluasi menggunakan berbagai metrik seperti Accuracy, Precision, Recall, dan F1-Score.

### **Langkah 4: Evaluasi**

```python
from sklearn.metrics import classification_report, roc_auc_score

# Melakukan prediksi pada data uji
y_pred = rf.predict(X_te)

# Menghitung probabilitas kelas churn (kelas 1)
y_prob = rf.predict_proba(X_te)[:, 1]

# Menampilkan hasil evaluasi
print("Classification Report:")
print(classification_report(y_te, y_pred))

# Menampilkan nilai ROC-AUC
roc_auc = roc_auc_score(y_te, y_prob)
print(f"ROC-AUC Score: {roc_auc:.3f}")
```
